## Problem 2: Data Filtering

**1. Passengers in first class (Pclass == 1)**

```python
first_class = df[df["Pclass"] == 1]
print("Count:", len(first_class))
print("Shape:", first_class.shape)
first_class.head()
```

**2. Passengers younger than 18**

```python
young = df[df["Age"] < 18]
print("Count:", len(young))
young[["Name", "Age", "Sex", "Survived"]].head()
```

**3. Male passengers who survived**

```python
male_survived = df[(df["Sex"] == "male") & (df["Survived"] == 1)]
print("Count:", len(male_survived))
print("Shape:", male_survived.shape)
male_survived.head()
```

**4. Passengers who paid a fare greater than 100**

```python
high_fare = df[df["Fare"] > 100]
print("Count:", len(high_fare))
print("Shape:", high_fare.shape)
high_fare[["Name", "Pclass", "Fare"]].head()
```

**Note:** Only the first rows are shown to keep the notebook short. The count above each table tells how many rows matched. Passengers with a missing `Age` are not counted in the under-18 filter.

## Problem 3: Sorting and Aggregation

**1. The 10 passengers who paid the highest fares**

```python
top10 = df.sort_values(by="Fare", ascending=False).head(10)
top10[["Name", "Pclass", "Fare", "Survived"]]
```

**2. Average passenger age**

```python
print("Average age:", round(df["Age"].mean(), 2))
```

**3. Number of male and female passengers**

```python
df["Sex"].value_counts()
```

**4. Average fare for each passenger class**

```python
df.groupby("Pclass")["Fare"].mean()
```

**5. Survival rate for each passenger class**

```python
df.groupby("Pclass")["Survived"].mean()
```

## Final Laboratory Challenge: Titanic Dataset Analysis

**Q1. How many passengers are in the dataset?**

```python
total_passengers = len(df)
print("Duplicate PassengerId:", df["PassengerId"].duplicated().sum())
print("Total passengers:", total_passengers)
```

**Answer:** There are **891** passengers.

**Q2. What is the average passenger age?**

```python
print("Average age:", round(df["Age"].mean(), 2))
```

**Answer:** The average age is about **29.70 years**. This is calculated from 714 passengers, because 177 passengers have no age recorded.

**Q3. Which passenger class has the highest number of passengers?**

```python
class_counts = df["Pclass"].value_counts().sort_index()
print(class_counts)
print("Largest passenger class:", class_counts.idxmax())
```

**Answer:** **Class 3** has the most passengers (491). Class 1 has 216 and class 2 has 184.

**Q4. What is the average fare paid by male and female passengers?**

```python
df.groupby("Sex")["Fare"].mean().round(2)
```

**Answer:** Female passengers paid more on average (about **44.48**) than male passengers (about **25.52**).

**Q5. Which gender has a higher survival rate?**

```python
survival_by_sex = df.groupby("Sex")["Survived"].mean().round(3)
print(survival_by_sex)
```

**Answer:** **Females** have a much higher survival rate (about 74.2%) than males (about 18.9%).

**Q6. How does the survival rate differ across passenger classes?**

```python
survival_by_class = df.groupby("Pclass")["Survived"].mean().round(3)
print(survival_by_class)
```

**Answer:** The survival rate goes down as the class number goes up: class 1 is about 63.0%, class 2 is about 47.3%, and class 3 is about 24.2%.
